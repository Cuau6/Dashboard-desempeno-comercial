## Paso 3: Diseño y planificación del dashboard

Hice la planeación antes de construir y la fui ajustando mientras avanzaba el dashboard.

**Qué pregunta responde cada vista**

- Vista 1 (Overview): ¿cómo está evolucionando el desempeño comercial de Andes Retail Group durante 2024-2025? Da una visión general y rápida del negocio con los principales indicadores, la evolución de los ingresos y las diferencias entre países, categorías y segmentos de clientes.
- Vista 2 (Detalle): ¿qué categorías, segmentos, países y regiones explican las diferencias en el desempeño y dónde hay oportunidades de mejora? Aquí se analizan con más detalle los resultados del Overview: ingresos, ganancia, margen, unidades vendidas y pedidos.

**KPIs y por qué**

- Ingresos totales: el valor de las ventas generadas.
- Ganancia: el resultado económico después de costos.
- Margen de ganancia (%): qué tan rentable es el negocio en relación con los ingresos.
- Unidades vendidas: el volumen comercializado.
- Número de pedidos: el nivel de actividad comercial.
- Ticket promedio: cuánto ingreso genera, en promedio, cada pedido.

**Gráficos elegidos**

- Evolución de los ingresos 2024-2025: líneas, porque muestran tendencias y variaciones mensuales.
- Ingresos por país: barras horizontales, para comparar directamente Perú, Chile y Colombia.
- Impacto de las categorías de producto: barras de ingresos por categoría, con ganancia, margen y unidades como información adicional.
- Segmentos de clientes: barras, para comparar ingresos y rentabilidad de cada segmento.
- Vista de detalle: tabla con país, región, categoría y sus principales indicadores.

**Jerarquía visual y distribución**

El orden es de lo general a lo específico, de arriba hacia abajo: visión general, evolución, comparación y diagnóstico. Arriba van el título y los segmentadores principales; luego las tarjetas de KPIs; en el centro, la evolución mensual de ingresos; después los gráficos por país y categoría; y al final los segmentos de clientes y, en la vista de detalle, las tablas para profundizar.

**Filtros**

Año (para comparar 2024 y 2025), fecha (periodos específicos), país (Perú, Chile y Colombia), región, categoría de producto (Electrónica, Ropa, Deportes y Hogar), segmento de cliente y estación (para ver patrones estacionales, considerando las estaciones del hemisferio sur).

## Paso 5: Narrativa con el modelo SCQA y mensaje asincrónico

Escribí la narrativa del dashboard con el modelo SCQA (Situación, Complicación, Pregunta y Respuesta): primero lo que se ve en el Overview y después los hallazgos de la vista de detalle. Al final resumí el insight principal en un mensaje corto para Slack.

#### Vista general (Overview)

**S (Situación):** Andes Retail Group vendió $5.5M en 2024-2025 (5,000 pedidos y 57.6K unidades), con una ganancia de $1.94M y un margen de 35.1 %. Opera en Perú (39 % de los ingresos), Chile (37 %) y Colombia (24 %), con cuatro categorías muy parejas (24 a 26 % cada una).

**C (Complicación):** los ingresos cayeron 6.8 % de 2024 a 2025 ($2.86M a $2.67M) y el ticket promedio bajó de $1,123 a $1,089. El margen se mantuvo plano en 35 %. Además, la venta depende de pocos clientes: Premium genera 47 % de los ingresos con solo 30 % de los pedidos.

**Q (Pregunta):** ¿de dónde viene la caída y en qué parte del negocio conviene actuar primero?

**A (Respuesta):** el margen no es el problema, porque se mantiene en alrededor de 35 % en todos los países, categorías y segmentos. El reto está en el volumen y el ticket, sobre todo en el segmento Premium y en los mercados de Perú y Chile. La vista de detalle muestra dónde se concentra la caída.

#### Vista Detalle

**S (Situación):** la venta es muy estacional. Verano (dic-feb) concentra 40.5 % de los ingresos ($2.24M) e invierno solo 11.8 %. El número de pedidos es parecido entre estaciones (entre 1,200 y 1,300), lo que cambia es el ticket: $1,722 en verano contra $549 en invierno.

**C (Complicación):** la caída de 2025 se concentra en verano: -$203K (-16.6 %), más que la caída total del año (-$195K). Las otras tres estaciones sumadas crecieron $8K.

**Q (Pregunta):** ¿qué segmento, país y categoría explican la caída del verano?

**A (Respuesta):**
- Segmento: Premium explica cerca de 83 % de la caída (-$168K, -28 %), con 26 % menos pedidos.
- País: Perú (-23 %) y Chile (-17.5 %). Colombia casi no cambió.
- Categoría: Deportes (-26 %) y Ropa (-22 %). Hogar creció, así que vale la pena ver qué se hizo distinto ahí.

**Recomendación:** preparar un plan de reactivación de clientes Premium para el verano 2026, enfocado en Perú y Chile, con ofertas en Deportes y Ropa.
